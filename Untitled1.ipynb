```python
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.optimizers import SGD

# ===============================
# 数据加载
# ===============================
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# 归一化
train_images = train_images / 255.0
test_images = test_images / 255.0

# ===============================
# 任务一：显示 train_image[6000]
# ===============================
plt.imshow(train_images[6000], cmap='gray')
plt.title("train_image[6000]")
plt.axis('off')
plt.show()

# ===============================
# 任务二：ax1 ax2 ax3 ax4
# ===============================
fig, axes = plt.subplots(1, 4, figsize=(10, 3))

axes[0].imshow(train_images[1000], cmap='gray')
axes[0].set_title("ax1: image[1000]")
axes[0].axis('off')

axes[1].imshow(train_images[2000], cmap='gray')
axes[1].set_title("ax2: image[2000]")
axes[1].axis('off')

axes[2].imshow(train_images[3000], cmap='gray')
axes[2].set_title("ax3: image[3000]")
axes[2].axis('off')

axes[3].imshow(train_images[4000], cmap='gray')
axes[3].set_title("ax4: image[4000]")
axes[3].axis('off')

plt.show()

# ===============================
# 任务三：创建模型（隐藏层 256 → 128）
# ===============================
model = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation='relu'),   # 原 256 改为 128
    Dense(10, activation='softmax')
])

# 截图 model.summary()
model.summary()

# ===============================
# 任务四：使用 SGD 编译模型
# ===============================
model.compile(
    optimizer=SGD(learning_rate=0.01),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# ===============================
# 任务五：epochs=2
# ===============================
history_2 = model.fit(
    train_images,
    train_labels,
    epochs=2,
    validation_data=(test_images, test_labels)
)

# ===============================
# 任务五：epochs=5
# ===============================
history_5 = model.fit(
    train_images,
    train_labels,
    epochs=5,
    validation_data=(test_images, test_labels)
)
```
